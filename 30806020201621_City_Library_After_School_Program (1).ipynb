```python
import pandas as pd
import sqlite3
import numpy as np
conn = sqlite3.connect("/content/db")

#Task 1 بسم الله
#required 1
req1="""SELECT m.member_id, m.first_name, COUNT(c.checkout_id) AS checkout_count
FROM members m
LEFT JOIN checkouts c
ON m.member_id = c.member_id
GROUP BY m.member_id """
result = pd.read_sql_query(req1, conn)
print(result)
conn.close()
```

```python
#follow task 1 req 2
conn = sqlite3.connect("/content/db")
req2 = """
SELECT book_id, title, author
FROM books
WHERE author LIKE 'S%'
"""

result2 = pd.read_sql_query(req2, conn)
print(result2)
```

```python
#follow task 1 req 3
req3="""SELECT b.title, COUNT(c.checkout_id) AS count
FROM books b
JOIN checkouts c
ON b.book_id = c.book_id
GROUP BY b.title
ORDER BY count DESC
LIMIT 5 """
result3 = pd.read_sql_query(req3, conn)
print(result3)
```

```python
#follow task 1 req 4
req4="""SELECT m.member_id, m.first_name, COUNT(c.checkout_id) AS borrowed
FROM members m
JOIN checkouts c
ON m.member_id = c.member_id
GROUP BY m.member_id
ORDER BY borrowed DESC
LIMIT 10"""
result4 = pd.read_sql_query(req4, conn)
print(result4)
```

```python
#follow task 1 req 5
req5="""SELECT *
FROM checkouts
JOIN members
ON checkouts.member_id = members.member_id
WHERE neighborhood = 'Nasr City'
ORDER BY checkout_date DESC
LIMIT 10 OFFSET 10"""
result5 = pd.read_sql_query(req5, conn)
print(result5)
```

```python
data=data.drop_duplicates()
print(data.shape)
```

```python
#نعالجج الميسينج فاليو
columns_treat_by_mode = [
    "first_name",
    "last_name",
    "neighborhood",
    "membership_status",
    "join_date",
    "checkout_date",
    "return_date",
    "title",
    "author"
]

for col in columns_treat_by_mode:
    data[col] = data[col].fillna(data[col].mode()[0])

data["grade"] = data["grade"].fillna(data["grade"].median())
data["checkout_id"] = data["checkout_id"].fillna("No Checkout")
data["book_id"] = data["book_id"].fillna("No Book")

print(data.isna().sum())
```

```python
#نعالج مشاكل ال consistence
data["neighborhood"]=data["neighborhood"].str.strip().str.title()
data['membership_status']=data['membership_status'].str.strip().str.title()
print(data)
```

```python
#نشوف ال wrong ids
wrong_ids = [1104, 1150, 1201]
print(data[data["member_id"].isin(wrong_ids)])
data = data[~data["member_id"].isin(wrong_ids)]
print(data)
```
